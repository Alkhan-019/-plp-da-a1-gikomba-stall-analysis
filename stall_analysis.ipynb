### Task 2: Weekly Sales Performance Analysis

The business recorded total sales of KES 22,050 during Week 1 and KES 24,400 during Week 2.

This represents a positive week-on-week growth of 10.66%, indicating an improvement in business performance.

The increase suggests that the stall experienced stronger sales during the second week. The business should maintain effective sales strategies, monitor customer demand, and ensure sufficient stock availability to sustain this growth.

```python
def classify_day(amount, average):
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"
```

```python
average = sum(sales) / len(sales)

daily_classifications = []

for day, amount in enumerate(sales, start=1):
    category = classify_day(amount, average)
    daily_classifications.append(category)

    print(f"Day {day}: KES {amount:,.2f} - {category}")
```

```python
category_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

for category in daily_classifications:
    category_counts[category] += 1

print("\nDAILY SALES CLASSIFICATION")
print("==========================")

for category, count in category_counts.items():
    print(f"{category}: {count} days")
```

```python
def compare_weekdays_weekends(sales):

    weekdays = sales[:5] + sales[7:12]
    weekends = sales[5:7] + sales[12:14]

    weekday_average = sum(weekdays) / len(weekdays)
    weekend_average = sum(weekends) / len(weekends)

    difference = weekend_average - weekday_average

    percentage_difference = (
        difference / weekday_average
    ) * 100

    return weekday_average, weekend_average, percentage_difference
```

```python
weekday_avg, weekend_avg, percentage_diff = compare_weekdays_weekends(sales)

print("WEEKDAY VS WEEKEND PERFORMANCE")
print("=" * 40)

print(f"Average Weekday Sales: KES {weekday_avg:,.2f}")
print(f"Average Weekend Sales: KES {weekend_avg:,.2f}")

print(f"Percentage Difference: {percentage_diff:.2f}%")

if weekend_avg > weekday_avg:
    print("Business Insight: Weekends perform better.")
else:
    print("Business Insight: Weekdays perform better.")
```

```python
def analyze_inventory(stock):

    for product, details in stock.items():

        quantity = details["qty"]
        reorder_level = details["reorder"]
        cost = details["cost"]
        price = details["price"]

        profit = price - cost
        profit_margin = (profit / price) * 100

        if quantity <= reorder_level:
            status = "REORDER"
        else:
            status = "STOCK OK"

        print(f"\nProduct: {product}")
        print(f"Quantity: {quantity}")
        print(f"Profit per unit: KES {profit}")
        print(f"Profit Margin: {profit_margin:.2f}%")
        print(f"Stock Status: {status}")
        
```

This task combines sales statistics, weekly performance,

```python
# Calculate sales statistics
total_sales = sum(sales)
average_sales = total_sales / len(sales)
highest_sales = max(sales)
lowest_sales = min(sales)

# Calculate weekly performance
week_1_total, week_2_total = week_totals(sales)

growth = ((week_2_total - week_1_total) / week_1_total) * 100

# Calculate weekday and weekend performance
weekday_avg, weekend_avg, percentage_diff = compare_weekdays_weekends(sales)

# Identify products requiring restocking
reorder_products = []

for product, details in stock.items():
    if details["qty"] <= details["reorder"]:
        reorder_products.append(product)

# Generate formatted report
print("=" * 55)
print("           GIKOMBA MARKET STALL PERFORMANCE REPORT")
print("=" * 55)

print("\n1. SALES PERFORMANCE")
print("-" * 55)
print(f"Total Sales:          KES {total_sales:,.2f}")
print(f"Average Daily Sales:  KES {average_sales:,.2f}")
print(f"Highest Sales:        KES {highest_sales:,.2f}")
print(f"Lowest Sales:         KES {lowest_sales:,.2f}")

print("\n2. WEEKLY PERFORMANCE")
print("-" * 55)
print(f"Week 1 Sales:         KES {week_1_total:,.2f}")
print(f"Week 2 Sales:         KES {week_2_total:,.2f}")
print(f"Weekly Growth:        {growth:.2f}%")

print("\n3. WEEKDAY VS WEEKEND")
print("-" * 55)
print(f"Weekday Average:      KES {weekday_avg:,.2f}")
print(f"Weekend Average:      KES {weekend_avg:,.2f}")
print(f"Weekend Difference:   {percentage_diff:.2f}%")

print("\n4. INVENTORY ALERTS")
print("-" * 55)

if reorder_products:
    for product in reorder_products:
        print(f"RESTOCK REQUIRED: {product}")
else:
    print("All products have sufficient stock.")

print("\n5. DAILY SALES CLASSIFICATION")
print("-" * 55)

for category, count in category_counts.items():
    print(f"{category}: {count} days")

print("\n" + "=" * 55)
print("                 END OF REPORT")
print("=" * 55)
```
